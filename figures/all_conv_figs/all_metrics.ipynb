```python
brain_mask_small = nib.load('/home/macke/mgloeckler90/dmri/data/data/nodif_brain_mask.nii.gz').get_fdata().astype(bool)
brain_mask_long = nib.load('/home/macke/mgloeckler90/dmri/data/new_data/nodif_brain_mask.nii.gz').get_fdata().astype(bool)
```

```python
from scipy.ndimage import binary_erosion
brain_mask_small = binary_erosion(brain_mask_small, iterations=2)
brain_mask_long = binary_erosion(brain_mask_long, iterations=2)
```

```python
def exclude_outliers(ksd, q=0.99):
    if ksd is None:
        return None
    threshold = np.quantile(ksd[brain_mask_small], q)
    ksd_filtered = np.copy(ksd)
    ksd_filtered[ksd > threshold] = threshold
    return ksd_filtered

def exclude_outliers_syn(ksd, q=0.99):
    if ksd is None:
        return None
    threshold = np.quantile(ksd, q)
    ksd_filtered = np.copy(ksd)
    ksd_filtered[ksd > threshold] = threshold
    return ksd_filtered
```

```python
# Summary stats (median, 10th, 90th percentiles)

def qstats(arr, mask=None):
    if arr is None:
        return np.nan, np.nan, np.nan
    data = arr[mask] if mask is not None else arr
    if data.size == 0:
        return np.nan, np.nan, np.nan
    data = data[data > 0]
    median = np.nanmedian(data)
    q10, q90 = np.nanquantile(data, [0.1, 0.9])
    return median, q10, q90

rows = []

# Synthetic
#ess_median, ess_q10, ess_q90 = qstats(syn_eval)
#rows.append({"dataset": "eval", "metric": "ESS", "median": ess_median, "q10": ess_q10, "q90": ess_q90})

#ess_long_median, ess_long_q10, ess_long_q90 = qstats(syn_eval_long)
#rows.append({"dataset": "long", "metric": "ESS", "median": ess_long_median, "q10": ess_long_q10, "q90": ess_long_q90})

#ksd_filtered_syn = exclude_outliers_syn(syn_ksd_short)
#ksd_median, ksd_q10, ksd_q90 = qstats(ksd_filtered_syn)
#rows.append({"dataset": "eval", "metric": "KSD", "median": ksd_median, "q10": ksd_q10, "q90": ksd_q90})

#ksd_long_filtered_syn = exclude_outliers_syn(syn_ksd_long)
#ksd_long_median, ksd_long_q10, ksd_long_q90 = qstats(ksd_long_filtered_syn)
#rows.append({"dataset": "long", "metric": "KSD", "median": ksd_long_median, "q10": ksd_long_q10, "q90": ksd_long_q90})

rmse_short_median, rmse_short_q10, rmse_short_q90 = qstats(syn_rmse_short)
rows.append({"dataset": "eval", "metric": "RMSE", "median": rmse_short_median, "q10": rmse_short_q10, "q90": rmse_short_q90})

#rmse_long_median, rmse_long_q10, rmse_long_q90 = qstats(syn_rmse_long)
#rows.append({"dataset": "long", "metric": "RMSE", "median": rmse_long_median, "q10": rmse_long_q10, "q90": rmse_long_q90})

# Real (masked)
#real_ess_bss_median, real_ess_bss_q10, real_ess_bss_q90 = qstats(real_ess_bss, brain_mask_small)
#rows.append({"dataset": "bss", "metric": "ESS", "median": real_ess_bss_median, "q10": real_ess_bss_q10, "q90": real_ess_bss_q90})

real_ksd_bss_filtered = exclude_outliers(real_ksd_bss)
real_ksd_noddi_filtered = exclude_outliers(real_ksd_noddi)
real_ksd_TTssN_filtered = exclude_outliers(real_ksd_TTssN)

real_ksd_bss_median, real_ksd_bss_q10, real_ksd_bss_q90 = qstats(real_ksd_bss_filtered, brain_mask_small)
rows.append({"dataset": "bss", "metric": "KSD", "median": real_ksd_bss_median, "q10": real_ksd_bss_q10, "q90": real_ksd_bss_q90})

real_ksd_noddi_median, real_ksd_noddi_q10, real_ksd_noddi_q90 = qstats(real_ksd_noddi_filtered, brain_mask_small)
rows.append({"dataset": "noddi", "metric": "KSD", "median": real_ksd_noddi_median, "q10": real_ksd_noddi_q10, "q90": real_ksd_noddi_q90})

real_ksd_TTssN_median, real_ksd_TTssN_q10, real_ksd_TTssN_q90 = qstats(real_ksd_TTssN_filtered, brain_mask_small)
rows.append({"dataset": "TTssN", "metric": "KSD", "median": real_ksd_TTssN_median, "q10": real_ksd_TTssN_q10, "q90": real_ksd_TTssN_q90})

real_rmse_bss_median, real_rmse_bss_q10, real_rmse_bss_q90 = qstats(real_rmse_bss, brain_mask_small)
rows.append({"dataset": "bss", "metric": "RMSE", "median": real_rmse_bss_median, "q10": real_rmse_bss_q10, "q90": real_rmse_bss_q90})

real_rmse_noddi_median, real_rmse_noddi_q10, real_rmse_noddi_q90 = qstats(real_rmse_noddi, brain_mask_small)
rows.append({"dataset": "noddi", "metric": "RMSE", "median": real_rmse_noddi_median, "q10": real_rmse_noddi_q10, "q90": real_rmse_noddi_q90})

real_rmse_TTssN_median, real_rmse_TTssN_q10, real_rmse_TTssN_q90 = qstats(real_rmse_TTssN, brain_mask_small)
rows.append({"dataset": "TTssN", "metric": "RMSE", "median": real_rmse_TTssN_median, "q10": real_rmse_TTssN_q10, "q90": real_rmse_TTssN_q90})

summary_df = pd.DataFrame(rows)
summary_df
```
